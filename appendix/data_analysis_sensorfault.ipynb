https://www.kaggle.com/datasets/arashnic/sensor-fault-detection-data

Time series of measurements on sensors uniquely identified by a Sensor Id.
During the serie of measurement the sensor is disconnected or on failure.

SensorID = 1 = PT100 temperature sensor, in an industrial environment, with dust & vibrations

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python
df = pd.read_csv('data/archive/sensorfault/sensor-fault-detection.csv',sep=";")
```

```python
df.head()
```

```python
print(f"Sensor fault detection data:")
print(f"--------------------------------")
print(f"Shape: {df.shape}\n")
print(f"Columns: {df.columns}\n")
print(f"Max-Min values")
print(f"Timestamp: max: {df.Timestamp.max()}, min: {df.Timestamp.min()}")
print(f"Value: max: {df.Value.max()}, min: {df.Value.min()}\n")

print(f"Empty values:\n{df.isna().sum()}")
```

```python
ax = sns.histplot(df, x='Value', hue='SensorId', legend=False)
ax.set(title=f'Value histogram')
plt.show()
```
